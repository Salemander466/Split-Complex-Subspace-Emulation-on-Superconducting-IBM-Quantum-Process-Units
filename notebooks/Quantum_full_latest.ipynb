```python
import numpy as np
import json
import time
from math import pi
from qiskit import QuantumCircuit, QuantumRegister, ClassicalRegister, transpile
from qiskit.transpiler.preset_passmanagers import generate_preset_pass_manager
from qiskit_ibm_runtime import QiskitRuntimeService, SamplerV2
from google.colab import userdata

# ============================================================
# Split-Complex Born Rule Deviation Test
# Purpose:
# Compare ordinary complex quantum phase behavior against the
# split-complex mapped operator across Z, X, and Y measurement bases.
#
# Important:
# This test does NOT prove new physics by itself.
# It tests whether the hardware result deviates from ordinary
# complex Hilbert-space predictions after transpilation.
# ============================================================

class SplitComplexBornRuleDeviationTest:
    def __init__(
        self,
        backend_name=None,
        shots=10000,
        optimization_level=3,
        theta_degrees=None,
        anti_cancel=True,
        seed=166,
    ):
        self.token = userdata.get("ibm")
        if not self.token:
            raise ValueError("Missing IBM token in Colab secrets under key: ibm")

        self.service = QiskitRuntimeService(
            channel="ibm_quantum_platform",
            token=self.token
        )

        if backend_name is None:
            self.backend = self.service.least_busy(
                simulator=False,
                operational=True
            )
        else:
            self.backend = self.service.backend(backend_name)

        self.shots = int(shots)
        self.optimization_level = int(optimization_level)
        self.anti_cancel = bool(anti_cancel)
        self.seed = int(seed)
        self.rng = np.random.default_rng(self.seed)

        if theta_degrees is None:
            self.theta_degrees = list(range(0, 181, 15))
        else:
            self.theta_degrees = theta_degrees

        print("Connected backend:", self.backend.name)
        print("Shots per circuit:", self.shots)
        print("Optimization level:", self.optimization_level)
        print("Theta sweep:", self.theta_degrees)
        print("Anti-cancellation enabled:", self.anti_cancel)

    def add_bell_state(self, qc, q):
        qc.h(q[0])
        qc.cx(q[0], q[1])

    def add_split_complex_mapped_gate(self, qc, q, theta):
        # Mapped split-complex-style reversible operator:
        # CX -> RY(2 theta) -> CX
        qc.cx(q[0], q[1])
        qc.ry(2.0 * theta, q[0])
        qc.cx(q[0], q[1])

    def add_basis_rotation(self, qc, q, basis):
        # Measure both qubits in selected basis.
        # Z: no rotation
        # X: H before measurement
        # Y: Sdg then H before measurement
        if basis == "Z":
            return
        elif basis == "X":
            qc.h(q[0])
            qc.h(q[1])
        elif basis == "Y":
            qc.sdg(q[0])
            qc.sdg(q[1])
            qc.h(q[0])
            qc.h(q[1])
        else:
            raise ValueError("Basis must be one of: Z, X, Y")

    def build_standard_complex_circuit(self, theta, basis):
        q = QuantumRegister(2, "q")
        c = ClassicalRegister(2, "c")
        qc = QuantumCircuit(q, c, name="standard_complex")

        self.add_bell_state(qc, q)

        # Ordinary complex phase baseline.
        # This is a standard unitary phase perturbation on one half of a Bell pair.
        qc.rz(theta, q[0])

        if self.anti_cancel:
            eps1 = float(self.rng.uniform(-0.01, 0.01))
            eps2 = float(self.rng.uniform(-0.01, 0.01))
            qc.ry(eps1, q[1])
            qc.rz(eps2, q[0])

        self.add_basis_rotation(qc, q, basis)
        qc.measure(q, c)
        return qc

    def build_split_complex_circuit(self, theta, basis):
        q = QuantumRegister(2, "q")
        c = ClassicalRegister(2, "c")
        qc = QuantumCircuit(q, c, name="split_complex_mapped")

        self.add_bell_state(qc, q)

        self.add_split_complex_mapped_gate(qc, q, theta)

        if self.anti_cancel:
            eps1 = float(self.rng.uniform(-0.01, 0.01))
            eps2 = float(self.rng.uniform(-0.01, 0.01))
            qc.rz(eps1, q[0])
            qc.ry(eps2, q[1])
            qc.sx(q[0])

        self.add_basis_rotation(qc, q, basis)
        qc.measure(q, c)
        return qc

    def counts_to_probs(self, counts):
        total = sum(counts.values())
        if total == 0:
            return {"00": 0.0, "01": 0.0, "10": 0.0, "11": 0.0}
        return {k: counts.get(k, 0) / total for k in ["00", "01", "10", "11"]}

    def parity_expectation(self, probs):
        # <ZZ> style parity score for whichever measurement basis was used.
        return (
            probs.get("00", 0.0)
            + probs.get("11", 0.0)
            - probs.get("01", 0.0)
            - probs.get("10", 0.0)
        )

    def summarize_circuit(self, circuit):
        return {
            "depth": circuit.depth(),
            "size": circuit.size(),
            "ops": dict(circuit.count_ops())
        }

    def run(self):
        bases = ["Z", "X", "Y"]
        raw_circuits = []
        metadata = []

        for theta_deg in self.theta_degrees:
            theta = np.deg2rad(theta_deg)

            for basis in bases:
                qc_standard = self.build_standard_complex_circuit(theta, basis)
                raw_circuits.append(qc_standard)
                metadata.append({
                    "model": "standard_complex",
                    "theta_deg": theta_deg,
                    "basis": basis
                })

                qc_split = self.build_split_complex_circuit(theta, basis)
                raw_circuits.append(qc_split)
                metadata.append({
                    "model": "split_complex_mapped",
                    "theta_deg": theta_deg,
                    "basis": basis
                })

        print("Raw circuits built:", len(raw_circuits))

        pass_manager = generate_preset_pass_manager(
            backend=self.backend,
            optimization_level=self.optimization_level
        )

        isa_circuits = []
        audits = []

        print("Transpiling circuits...")
        for idx, qc in enumerate(raw_circuits):
            isa = pass_manager.run(qc)
            isa_circuits.append(isa)
            audits.append(self.summarize_circuit(isa))

        print("Submitting to IBM backend:", self.backend.name)
        sampler = SamplerV2(self.backend)
        job = sampler.run(isa_circuits, shots=self.shots)

        print("Job ID:", job.job_id())
        result = job.result()

        rows = []

        for i, item in enumerate(metadata):
            counts = result[i].data.c.get_counts()
            probs = self.counts_to_probs(counts)
            parity = self.parity_expectation(probs)

            row = {
                "backend": self.backend.name,
                "job_id": job.job_id(),
                "model": item["model"],
                "theta_deg": item["theta_deg"],
                "basis": item["basis"],
                "shots": self.shots,
                "counts": counts,
                "p00": probs["00"],
                "p01": probs["01"],
                "p10": probs["10"],
                "p11": probs["11"],
                "same_state_probability": probs["00"] + probs["11"],
                "opposite_state_probability": probs["01"] + probs["10"],
                "parity_expectation": parity,
                "compiled_depth": audits[i]["depth"],
                "compiled_size": audits[i]["size"],
                "compiled_ops": audits[i]["ops"],
            }
            rows.append(row)

        comparison = self.compare_models(rows)

        output = {
            "test_name": "Split-Complex Born Rule Deviation Test",
            "backend": self.backend.name,
            "shots": self.shots,
            "optimization_level": self.optimization_level,
            "anti_cancel": self.anti_cancel,
            "rows": rows,
            "comparison": comparison,
        }

        filename = "split_complex_born_rule_deviation_results.json"
        with open(filename, "w") as f:
            json.dump(output, f, indent=2)

        self.print_report(rows, comparison, filename)
        return output

    def compare_models(self, rows):
        # Compare standard vs split result for each theta and basis.
        # This does not prove new physics. It identifies where the hardware
        # behavior diverges from the ordinary phase baseline.
        grouped = {}

        for r in rows:
            key = (r["theta_deg"], r["basis"])
            grouped.setdefault(key, {})
            grouped[key][r["model"]] = r

        comparisons = []

        for key, pair in grouped.items():
            theta_deg, basis = key
            if "standard_complex" not in pair or "split_complex_mapped" not in pair:
                continue

            a = pair["standard_complex"]
            b = pair["split_complex_mapped"]

            delta_parity = b["parity_expectation"] - a["parity_expectation"]
            delta_same = b["same_state_probability"] - a["same_state_probability"]

            # Conservative binomial-style standard error approximation.
            # Used only as a screening statistic, not a full physics claim.
            se = np.sqrt(
                max(1e-12, a["same_state_probability"] * (1 - a["same_state_probability"]) / self.shots)
                +
                max(1e-12, b["same_state_probability"] * (1 - b["same_state_probability"]) / self.shots)
            )

            z_score_same = delta_same / se if se > 0 else 0.0

            comparisons.append({
                "theta_deg": theta_deg,
                "basis": basis,
                "standard_same_probability": a["same_state_probability"],
                "split_same_probability": b["same_state_probability"],
                "delta_same_probability": delta_same,
                "standard_parity": a["parity_expectation"],
                "split_parity": b["parity_expectation"],
                "delta_parity": delta_parity,
                "screening_z_score_same_probability": z_score_same,
            })

        return comparisons

    def print_report(self, rows, comparison, filename):
        print("\n============================================================")
        print("SPLIT-COMPLEX BORN RULE DEVIATION TEST REPORT")
        print("============================================================")
        print("Backend:", self.backend.name)
        print("Shots:", self.shots)
        print("Optimization level:", self.optimization_level)
        print("Anti-cancellation:", self.anti_cancel)
        print("Saved JSON:", filename)
        print("------------------------------------------------------------")

        print("\nTop compiled circuit audits:")
        for r in rows[:6]:
            print(
                f"{r['model']} | theta={r['theta_deg']} | basis={r['basis']} | "
                f"depth={r['compiled_depth']} | size={r['compiled_size']} | ops={r['compiled_ops']}"
            )

        print("\nLargest standard-vs-split deviations by absolute z-score:")
        sorted_comp = sorted(
            comparison,
            key=lambda x: abs(x["screening_z_score_same_probability"]),
            reverse=True
        )

        for item in sorted_comp[:10]:
            print(
                f"theta={item['theta_deg']:>3} deg | basis={item['basis']} | "
                f"standard_same={item['standard_same_probability']:.4f} | "
                f"split_same={item['split_same_probability']:.4f} | "
                f"delta={item['delta_same_probability']:+.4f} | "
                f"z={item['screening_z_score_same_probability']:+.2f}"
            )

        print("\nInterpretation:")
        print("If deviations are small or simulator-reproducible, this supports emulation, not new physics.")
        print("If a deviation is large, repeat it across backends, optimization levels, and noise-model simulation.")
        print("A serious new-physics claim would require repeated non-simulator-reproducible deviations above 5 sigma.")

# ============================================================
# RUN CONFIGURATION
# ============================================================

test = SplitComplexBornRuleDeviationTest(
    backend_name=None,          # Use None for least-busy real backend, or set "ibm_marrakesh"
    shots=10000,                # Increase to 20000+ for stronger statistics
    optimization_level=3,
    theta_degrees=list(range(0, 181, 15)),
    anti_cancel=True,
    seed=166
)

results = test.run()
```
